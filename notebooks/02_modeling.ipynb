## Setup

```python
import pandas as pd
import numpy as np
from pathlib import Path

DATA_DIR = Path("../data/processed")
df_elo = pd.read_csv(DATA_DIR / "elo.csv")

standings = {team: 0 for team in df_elo["team"]}
teams = list(standings.keys())
```

```python
def win_prob(elo_a, elo_b):
    return 1 / (1 + 10 ** ((elo_b - elo_a) / 400))

    
fixtures = []
for i, home in enumerate(teams):
    for j, away in enumerate(teams):
        if i < j:  # her eşleşme bir kez
            fixtures.append((home, away))

len(fixtures)
```

## Baseline model

```python
fixtures = []
for i, home in enumerate(teams):
    for j, away in enumerate(teams):
        if i < j:  # her eşleşme bir kez
            fixtures.append((home, away))

len(fixtures)
```

```python
n_sim = 10000
points = {team: np.zeros(n_sim) for team in teams}

for sim in range(n_sim):
    elo_sim = dict(zip(df_elo["team"], df_elo["elo"]))

    for home, away in fixtures:
        p_home = win_prob(elo_sim[home], elo_sim[away])
        rnd = np.random.rand()

        if rnd < p_home:
            points[home][sim] += 3
        else:
            points[away][sim] += 3

final_avg = {team: np.mean(points[team]) for team in teams}
pd.DataFrame({"team": list(final_avg.keys()), "avg_points": list(final_avg.values())}).sort_values("avg_points", ascending=False)
```
